```python
# %% [markdown]
# # 📊 Capítulo 3: Momentum Crash
# *Vender cuando todos son eufóricos — Protección contra burbujas especulativas*
# 
# **Estrategia:** Identificar sectores ETFs en euforia (top momentum + overvaluation) y ROTAR hacia activos defensivos (bonos, oro, cash). Holding period de 3 meses (rápido, para reaccionar a cambios de régimen).
# 
# **Universo:** Mismos 19 ETFs del Capítulo 2, pero con lógica INVERSA.
# 
# **Objetivo:** Proteger capital durante burbujas especulativas. No ganar MÁS que el mercado, sino perder MENOS durante los crashes. Alpha positivo = menos dolor en crisis.

# %%
# ==============================================================================
# BLOQUE 0: CONFIGURACIÓN DE LA ESTRATEGIA — MOMENTUM CRASH
# ==============================================================================

# --- Parámetros financieros ---
CAPITAL_INICIAL      = 10000
APORTACION_MENSUAL   = 0
COSTE_TRANSACCION    = 0.0015       # 0.15% por operación (ligeramente más alto)
FREQ_REBALANCEO      = 'M'          # Revisión mensual
BENCHMARK            = 'SPY'

# --- Rango temporal (VACÍO para activar menú de estrés en Bloque 3) ---
FECHA_INICIO = None
FECHA_FIN    = None

# ============================================================================
# PARÁMETROS ESPECÍFICOS DE MOMENTUM CRASH
# ============================================================================
TOP_N_GANADORES      = 5           # Cuántos "ganadores" excluir/reducir
UMBRAL_MOMENTUM      = 0.15        # Ganancia 12m > 15% (euforia moderada)
UMBRAL_RSI_OVER      = 70.0        # RSI(14) > 70 (sobrecompra)
HOLDING_MESES        = 3           # Rotación rápida (vs 6 meses del Contrarian)
CASH_TICKER          = 'SHV'       # Cash como refugio
SHORT_ALLOWED        = False       # False por defecto (corto difícil para retail)

# --- Ventana mínima necesaria ---
VENTANA_MINIMA_DIAS  = int(252 * 1.20)  # ≈ 302 días

# --- Rutas portables ---
RUTA_ENTRADA_R4 = None
RUTA_SALIDA_CSV = None

# --- Universo de activos (IDÉNTICO al Cap. 2 para comparabilidad) ---
UNIVERSO_MOMENTUM = {
    'broad_market': {
        'SPDR S&P 500':           {'isin': 'SPY',  'ticker': 'SPY'},
        'Invesco QQQ':            {'isin': 'QQQ',  'ticker': 'QQQ'},
        'Russell 2000 Small Cap': {'isin': 'IWM',  'ticker': 'IWM'},
    },
    'sectoriales_antiguos': {
        'Technology SPDR':        {'isin': 'XLK',  'ticker': 'XLK'},
        'Financials SPDR':        {'isin': 'XLF',  'ticker': 'XLF'},
        'Energy SPDR':            {'isin': 'XLE',  'ticker': 'XLE'},
        'Healthcare SPDR':        {'isin': 'XLV',  'ticker': 'XLV'},
        'Industrial SPDR':        {'isin': 'XLI',  'ticker': 'XLI'},
        'Consumer Staples SPDR':  {'isin': 'XLP',  'ticker': 'XLP'},
        'Utilities SPDR':         {'isin': 'XLU',  'ticker': 'XLU'},
        'Materials SPDR':         {'isin': 'XLB',  'ticker': 'XLB'},
        'Consumer Disc SPDR':     {'isin': 'XLY',  'ticker': 'XLY'},
    },
    'internacional_bonos': {
        'MSCI EAFE Developed':    {'isin': 'EFA',  'ticker': 'EFA'},
        'MSCI Emerging Markets':  {'isin': 'EEM',  'ticker': 'EEM'},
        'iShares 20+ Year Treasury': {'isin': 'TLT', 'ticker': 'TLT'},
    },
    'alternativos': {
        'Gold ETF':               {'isin': 'GLD',  'ticker': 'GLD'},
    },
    'cash': {
        'Short Treasury ETF':    {'isin': 'SHV',  'ticker': 'SHV'},
    },
}

print("✅ Configuración Momentum Crash cargada:")
print(f"   • Capital inicial: {CAPITAL_INICIAL:,.0f} €")
total_etfs = len([a for cat in UNIVERSO_MOMENTUM.values() for a in cat])
print(f"   • Universo: {total_etfs} ETFs (idéntico al Cap. 2)")
print(f"   • Top ganadores a identificar: {TOP_N_GANADORES}")
print(f"   • Filtro momentum: ≥ {UMBRAL_MOMENTUM:.0%} en 12 meses")
print(f"   • Filtro RSI (sobrecompra): ≥ {UMBRAL_RSI_OVER}")
print(f"   • Holding period: {HOLDING_MESES} meses (rotación rápida)")
print(f"   • Cash proxy: {CASH_TICKER}")
print(f"   • Benchmark: {BENCHMARK}")
print(f"   • Ventana mínima requerida: {VENTANA_MINIMA_DIAS} días")

# %%
# ==============================================================================
# BLOQUE 1: ENTORNO E IMPORTACIONES
# ==============================================================================

# --- Aviso legal ---
print("="*70)
print("⚠️  AVISO LEGAL")
print("="*70)
print("""
Este notebook tiene fines EXCLUSIVAMENTE EDUCATIVOS y de investigación
cuantitativa. NO constituye asesoramiento financiero ni recomendación de
inversión. El rendimiento pasado no garantiza resultados futuros.
""")
print("="*70)

# --- Detección de entorno ---
import sys
import os
import warnings
warnings.filterwarnings('ignore')

IN_COLAB = "google.colab" in sys.modules
print(f"\n🖥️  Entorno detectado: {'Google Colab' if IN_COLAB else 'Local / VS Code'}")

# --- Importaciones estándar ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
from pathlib import Path

# --- Búsqueda dinámica de la carpeta 'core' ---
def buscar_core():
    """Busca la carpeta 'core' subiendo desde el notebook."""
    try:
        import ipykernel
        try:
            from jupyter_server import serverapp as app_module
        except ImportError:
            from notebook import notebookapp as app_module
        import json, re, urllib.request
        connection_file = os.path.basename(ipykernel.get_connection_file())
        match = re.search(r'kernel-(.*)\.json', connection_file)
        if match:
            kernel_id = match.group(1)
            for srv in app_module.list_running_servers():
                try:
                    url = srv['url'] + 'api/sessions'
                    token = srv.get('token', '')
                    headers = {'Authorization': f'token {token}'} if token else {}
                    req = urllib.request.Request(url, headers=headers)
                    with urllib.request.urlopen(req, timeout=2) as resp:
                        sessions = json.loads(resp.read())
                    for sess in sessions:
                        if sess.get('kernel', {}).get('id') == kernel_id:
                            ruta_nb = os.path.join(srv['root_dir'], sess['notebook']['path'])
                            ruta_dir = os.path.dirname(os.path.abspath(ruta_nb))
                            actual = Path(ruta_dir)
                            for parent in [actual] + list(actual.parents):
                                if (parent / 'core').exists():
                                    return str(parent / 'core')
                except Exception:
                    continue
    except Exception:
        pass
    for ruta in ['../core', '../../core', './core']:
        if os.path.isdir(ruta):
            return os.path.abspath(ruta)
    return None

if "google.colab" in sys.modules:
    import subprocess
    _REPO = "/content/Estrategias_No_Convencionales"
    if not os.path.isdir(os.path.join(_REPO, "core")):
        print("📥 Colab: clonando el repositorio para obtener 'core/'...")
        subprocess.run(["git", "clone", "--depth", "1",
                        "https://github.com/akitxu/Estrategias_No_Convencionales.git", _REPO], check=True)
    os.chdir(os.path.join(_REPO, "notebooks"))

RUTA_CORE = buscar_core()
if RUTA_CORE and os.path.isdir(RUTA_CORE):
    sys.path.insert(0, os.path.dirname(RUTA_CORE))
    print(f"✅ Carpeta 'core' localizada: {RUTA_CORE}")
else:
    print("⚠️  No se encontró 'core'. Asegúrate de que la estructura del proyecto es correcta.")

# --- Importación segura de módulos ---
try:
    from core import (
        GestorProyecto,
        GestorDatos,
        MotorBacktest,
        MotorBacktestDinamico,
        CalculadorMetricas,
        CalculadorMetricasExtras,
        GeneradorInformes
    )
    print("✅ Módulos del motor importados correctamente")
except ImportError as e:
    print(f"❌ Error importando módulos: {e}")
    raise

# --- Asignación automática de rutas ---
BASE_DIR = os.path.dirname(RUTA_CORE) if RUTA_CORE else os.getcwd()
print(f"\n📂 BASE_DIR del proyecto: {BASE_DIR}")

# %%
# ==============================================================================
# BLOQUE 2: LIMPIEZA DE DIRECTORIOS
# ==============================================================================

gestor_proyecto = GestorProyecto(base_dir_manual=BASE_DIR)
gestor_proyecto.imprimir_resumen()
gestor_proyecto.menu_limpieza_directorios()

# %%
# ==============================================================================
# BLOQUE 3: GESTIÓN DE FECHAS (MENÚ DE ESTRÉS)
# ==============================================================================

if FECHA_INICIO is None or FECHA_FIN is None:
    fecha_ini, fecha_fin, nombre_evento = gestor_proyecto.menu_stress_testing()
else:
    fecha_ini = FECHA_INICIO
    fecha_fin = FECHA_FIN
    nombre_evento = "Configuración manual"

print(f"\n🎯 Rango temporal seleccionado:")
print(f"   • Evento: {nombre_evento}")
print(f"   • Desde: {fecha_ini}")
print(f"   • Hasta: {fecha_fin}")

# %%
# ==============================================================================
# BLOQUE 4: MENÚ DEL GESTOR DE DATOS
# ==============================================================================

print("\n" + "="*70)
print("📊 INICIALIZANDO GESTOR DE DATOS PARA MOMENTUM CRASH")
print("="*70)

if not UNIVERSO_MOMENTUM or len(UNIVERSO_MOMENTUM) == 0:
    print("❌ ERROR: UNIVERSO_MOMENTUM está vacío.")
    continuar = False
elif len([a for cat in UNIVERSO_MOMENTUM.values() for a in cat]) == 0:
    print("❌ ERROR: No hay activos dentro de las categorías.")
    continuar = False
else:
    total_activos = len([a for cat in UNIVERSO_MOMENTUM.values() for a in cat])
    print(f"✅ Universo válido con {total_activos} activos.")
    
    try:
        gestor = GestorDatos(
            diccionario_datos=UNIVERSO_MOMENTUM,
            fecha_inicio=fecha_ini,
            fecha_fin=fecha_fin,
            gestor_proyecto=gestor_proyecto
        )
        
        print(f"\n📅 GestorDatos inicializado:")
        print(f"   • Período: {fecha_ini} → {fecha_fin}")
        print(f"   • Evento: {nombre_evento}")
        
        print("\n📥 Ejecutando menú de importación/descarga...")
        continuar = gestor.menu()
        
        if continuar and gestor.precios_ajustados is not None:
            print(f"\n✅ Datos preparados para el motor.")
            print(f"   • Columnas cargadas: {len(gestor.precios_ajustados.columns)}")
        else:
            print("\n⛔ Usuario canceló o hubo error. Abortando.")
            
    except Exception as e:
        print(f"\n❌ ERROR: {type(e).__name__}: {e}")
        raise

# %%
# ==============================================================================
# BLOQUE 5: GENERADOR DE PESOS FLEXIBLE PARA MOMENTUM CRASH
# ==============================================================================

print("\n" + "="*70)
print("🔧 GENERADOR DE PESOS FLEXIBLE (*ARGS)")
print("="*70)

def generador_pesos_final(*args):
    """
    Generador flexible que acepta cualquier número de argumentos.
    El motor pasa 4 args: (fecha, datos, rentabilidades, contexto)
    """
    num_args = len(args)
    
    if num_args >= 4:
        fecha = args[0]
        datos = args[1]
        rentabilidades = args[2]
        contexto = args[3]
    elif num_args >= 3:
        fecha = args[0]
        datos = args[1]
        contexto = args[2]
    elif num_args >= 2:
        fecha = args[0]
        datos = args[1]
        contexto = None
    else:
        return ({CASH_TICKER: 1.0}, {})
    
    # Extraer precios históricos
    if isinstance(datos, dict):
        precios_hist = datos.get('precios', datos.get('data', datos))
    elif hasattr(datos, 'loc'):
        if fecha is not None and fecha in datos.index:
            precios_hist = datos.loc[:fecha]
        elif fecha is not None:
            idx = datos.index.searchsorted(fecha)
            precios_hist = datos.iloc[:idx] if idx > 0 else datos
        else:
            precios_hist = datos
    else:
        if 'gestor' in globals():
            precios_hist = gestor.precios_ajustados.loc[:fecha] if fecha else gestor.precios_ajustados
        else:
            return ({CASH_TICKER: 1.0}, {})
    
    if precios_hist is None or precios_hist.empty:
        return ({CASH_TICKER: 1.0}, {})
    
    # LÓGICA DE MOMENTUM CRASH
    top_excluir = TOP_N_GANADORES
    cash_ticker = CASH_TICKER
    
    # Calcular momentum 12m
    momentum_scores = {}
    for col in precios_hist.columns:
        if col.upper() == cash_ticker.upper():
            continue
        serie = precios_hist[col].dropna()
        if len(serie) < 252:
            momentum_scores[col] = 0.0
        else:
            momentum_scores[col] = (serie.iloc[-1] / serie.iloc[-252]) - 1
    
    # Identificar sobrecomprados
    sorted_momentum = sorted(momentum_scores.items(), key=lambda x: x[1], reverse=True)
    sobrecomprados = [item[0] for item in sorted_momentum[:top_excluir]]
    
    # Asignar pesos
    pesos = {}
    activos_viables = [c for c in precios_hist.columns 
                       if c.upper() != cash_ticker.upper() and c not in sobrecomprados]
    
    if len(activos_viables) == 0:
        pesos[cash_ticker] = 1.0
    else:
        weight_per_active = 0.5 / len(activos_viables)
        for c in activos_viables:
            pesos[c] = weight_per_active
        pesos[cash_ticker] = 0.5
    
    if contexto is None:
        contexto = {}
    
    return (pesos, contexto)

generador = generador_pesos_final

print("✅ Generador configurado:")
print(f"   • Acepta: *args (cualquier número de argumentos)")
print(f"   • Devuelve: (weights_dict, contexto)")

# PRUEBA CON 4 ARGUMENTOS
print("\n🔬 Prueba con 4 argumentos (como el motor)...")
try:
    fecha_prueba = gestor.precios_ajustados.index[252]
    datos_prueba = gestor.precios_ajustados.loc[:fecha_prueba]
    renta_prueba = gestor.rentabilidades_diarias.loc[:fecha_prueba] if hasattr(gestor, 'rentabilidades_diarias') else None
    contexto_prueba = {'test': True}
    
    pesos, contexto_salida = generador(fecha_prueba, datos_prueba, renta_prueba, contexto_prueba)
    
    print(f"✅ Prueba exitosa!")
    print(f"   • Aceptó 4 argumentos correctamente")
    print(f"   • Peso en {CASH_TICKER}: {pesos.get(CASH_TICKER, 0):.1%}")
    print(f"   • Total activos en pesos: {len(pesos)}")
        
except Exception as e:
    print(f"❌ Error en prueba: {type(e).__name__}: {e}")
    raise

print("\n" + "="*70)
print("✅ GENERADOR LISTO")
print("="*70)

# %%
# ==============================================================================
# BLOQUE 6: EJECUCIÓN DEL MOTOR DE BACKTEST
# ==============================================================================

print("\n" + "="*70)
print("⏳ EJECUTANDO BACKTEST MOMENTUM CRASH")
print("="*70)

if 'generador' not in locals():
    print("❌ ERROR: 'generador' no definido")
    raise NameError("generador no definido")

print(f"✅ Generador: {generador.__name__}")

motor_dinamico = MotorBacktestDinamico(
    precios=gestor.precios_ajustados,
    rentabilidades=gestor.rentabilidades_diarias,
    generador_pesos=generador,
    frecuencia_decision=FREQ_REBALANCEO,
    coste_transaccion=COSTE_TRANSACCION,
    benchmark_ticker=BENCHMARK,
    permitir_short=SHORT_ALLOWED,
    max_short_porcentaje=0.0 if not SHORT_ALLOWED else 0.3
)

print(f"\n⚙️  Motor configurado:")
print(f"   • Frecuencia: {FREQ_REBALANCEO}")
print(f"   • Coste: {COSTE_TRANSACCION:.1%}")
print(f"   • Benchmark: {BENCHMARK}")

print("\n⏳ Ejecutando simulación...")
serie_patrimonio = motor_dinamico.ejecutar_backtest(
    capital_inicial=CAPITAL_INICIAL,
    estado_inicial={}
)

print("\n✅ Backtest completado.")
print(f"   • Patrimonio final: {serie_patrimonio.iloc[-1]:,.2f} €")
print(f"   • Rentabilidad: {((serie_patrimonio.iloc[-1] / CAPITAL_INICIAL) - 1) * 100:.2f}%")
print(f"   • Rebalanceos: {len(motor_dinamico.log_transacciones)}")

if len(motor_dinamico.log_transacciones) > 0:
    print(f"\n🎉 ÉXITO: {len(motor_dinamico.log_transacciones)} rebalanceos realizados.")
else:
    print(f"\n⚠️  CRÍTICO: 0 rebalanceos. Revisa las advertencias anteriores.")

# %%
# ==============================================================================
# BLOQUE 7: MÉTRICAS Y COMPARATIVA
# ==============================================================================

print("\n" + "="*70)
print("📊 ANÁLISIS COMPARATIVO: CONTRARIAN vs MOMENTUM CRASH")
print("="*70)

auditor = CalculadorMetricas(
    serie_patrimonio=motor_dinamico.serie_patrimonio,
    serie_buy_hold=motor_dinamico.serie_buy_hold,
    serie_benchmark=motor_dinamico.serie_benchmark
)
metricas_reb, metricas_bh = auditor.calcular_todas_las_metricas(rf=0.02)

print("\n📋 MÉTRICAS ESTÁNDAR:")
for k, v in metricas_reb.items():
    if isinstance(v, float):
        if 'Ratio' in k or 'Beta' in k or 'Alpha' in k:
            print(f"   • {k:<25}: {v:.2f}")
        elif abs(v) < 1:
            print(f"   • {k:<25}: {v:.2%}")
        else:
            print(f"   • {k:<25}: {v:,.2f}")

# Métricas de protección
print("\n" + "="*70)
print("🛡️  MÉTRICAS DE PROTECCIÓN")
print("="*70)

if motor_dinamico.serie_benchmark is not None and not motor_dinamico.serie_benchmark.empty:
    returns_s = motor_dinamico.serie_patrimonio.pct_change().dropna()
    returns_b = motor_dinamico.serie_benchmark.pct_change().dropna()
    
    common_idx = returns_s.index.intersection(returns_b.index)
    returns_s = returns_s.loc[common_idx]
    returns_b = returns_b.loc[common_idx]
    
    downside_months = returns_b[returns_b < 0]
    if len(downside_months) == 0:
        downside_capture = 0.0
    else:
        corresponding_s = returns_s.loc[downside_months.index]
        downside_capture = (corresponding_s.mean() / downside_months.mean()) * 100
    
    upside_months = returns_b[returns_b > 0]
    if len(upside_months) == 0:
        upside_capture = 0.0
    else:
        corresponding_s = returns_s.loc[upside_months.index]
        upside_capture = (corresponding_s.mean() / upside_months.mean()) * 100
    
    protection_ratio = upside_capture / downside_capture if downside_capture != 0 else 0
    
    protection_metrics = {
        'Downside Capture': downside_capture,
        'Upside Capture': upside_capture,
        'Protection Ratio': protection_ratio
    }
    
    print("\n🛡️  RESULTADOS DE PROTECCIÓN:")
    for k, v in protection_metrics.items():
        if isinstance(v, float):
            if 'Capture' in k:
                print(f"   • {k:<25}: {v:.1f}%")
            else:
                print(f"   • {k:<25}: {v:.2f}")
    
    print("\n📌 INTERPRETACIÓN:")
    if protection_metrics['Downside Capture'] < 50:
        print(f"   🟢 ¡Excelente! Pierdes MENOS del 50% que el mercado en crisis.")
    elif protection_metrics['Downside Capture'] < 80:
        print(f"   🟡 Bueno. Pierdes menos que el mercado.")
    else:
        print(f"   🔴 Preocupante. La estrategia no protege adecuadamente.")
else:
    protection_metrics = {'Downside Capture': 0.0, 'Upside Capture': 0.0, 'Protection Ratio': 0.0}

# Comparativa
print("\n" + "="*70)
print("📊 COMPARATIVA FINAL: CAP. 2 vs CAP. 3")
print("="*70)

mc_cagr = ((motor_dinamico.serie_patrimonio.iloc[-1] / CAPITAL_INICIAL) ** (1 / (len(motor_dinamico.serie_patrimonio)/252))) - 1
mc_sharpe = metricas_reb.get('Sharpe Ratio', 0)
mc_max_dd = (motor_dinamico.serie_patrimonio / motor_dinamico.serie_patrimonio.cummax() - 1).min()
mc_beta = metricas_reb.get('Beta', 0)
mc_rebalanceos = len(motor_dinamico.log_transacciones)

contrarian_data = {
    'CAGR': 6.25,
    'Sharpe': 0.48,
    'Max DD': -42.31,
    'Beta': 0.37,
    'Rebalanceos/año': 3.1
}

comparativa = pd.DataFrame([
    {
        'Métrica': 'CAGR',
        'Contrarian (Cap. 2)': f"{contrarian_data['CAGR']:.2f}%",
        'Momentum Crash (Cap. 3)': f"{mc_cagr*100:.2f}%",
        'Diferencia': f"{mc_cagr*100 - contrarian_data['CAGR']:+.2f}pp"
    },
    {
        'Métrica': 'Sharpe',
        'Contrarian (Cap. 2)': f"{contrarian_data['Sharpe']:.2f}",
        'Momentum Crash (Cap. 3)': f"{mc_sharpe:.2f}",
        'Diferencia': f"{mc_sharpe - contrarian_data['Sharpe']:+.2f}"
    },
    {
        'Métrica': 'Max Drawdown',
        'Contrarian (Cap. 2)': f"{abs(contrarian_data['Max DD']):.1f}%",
        'Momentum Crash (Cap. 3)': f"{abs(mc_max_dd*100):.1f}%",
        'Diferencia': f"{abs(mc_max_dd*100) - abs(contrarian_data['Max DD']):+.1f}pp"
    },
    {
        'Métrica': 'Beta',
        'Contrarian (Cap. 2)': f"{contrarian_data['Beta']:.2f}",
        'Momentum Crash (Cap. 3)': f"{mc_beta:.2f}",
        'Diferencia': f"{mc_beta - contrarian_data['Beta']:+.2f}"
    },
    {
        'Métrica': 'Rebalanceos/año',
        'Contrarian (Cap. 2)': f"{contrarian_data['Rebalanceos/año']:.1f}",
        'Momentum Crash (Cap. 3)': f"{mc_rebalanceos/19.6:.1f}",
        'Diferencia': f"{mc_rebalanceos/19.6 - contrarian_data['Rebalanceos/año']:+.1f}"
    },
    {
        'Métrica': 'Propósito',
        'Contrarian (Cap. 2)': 'Alpha defensivo',
        'Momentum Crash (Cap. 3)': 'Protección anticrisis',
        'Diferencia': '-'
    }
])

try:
    from IPython.display import display
    display(comparativa)
except ImportError:
    print(comparativa.to_string(index=False))

# Diagnóstico final
print("\n" + "="*70)
print("🩺 DIAGNÓSTICO FINAL MOMENTUM CRASH")
print("="*70)

anos_backtest = len(motor_dinamico.serie_patrimonio) / 252

print(f"\n   • Período: {anos_backtest:.1f} años")
print(f"   • Max DD: {mc_max_dd:.1%}")
print(f"   • CAGR: {mc_cagr:.2%}")
print(f"   • Rebalanceos/año: {mc_rebalanceos/max(anos_backtest, 1):.1f}")

if mc_max_dd < -0.25:
    print(f"   ⚠️  Max DD elevado para protección (mejorar parámetros)")
elif mc_max_dd < -0.20:
    print(f"   🟡 Max DD aceptable, mejorable")
else:
    print(f"   ✅ Max DD excelente para estrategia de protección")

if mc_cagr < 0.03:
    print(f"   ⚠️  CAGR bajo (esperado para 'seguro')")
elif mc_cagr < 0.06:
    print(f"   ✅ CAGR razonable para estrategia defensiva")
else:
    print(f"   🟢 CAGR excepcional para momentum crash")

print(f"\n💡 LECCIONES DEL CAPÍTULO 3:")
print(f"   • Momentum Crash: CAGR {mc_cagr:.2%}, DD {mc_max_dd:.1%}")
print(f"   • Contrarian Agresivo: CAGR {contrarian_data['CAGR']}%, DD {abs(contrarian_data['Max DD']):.1f}%")
print(f"   • Momentum protege MÁS pero gana MENOS")

# %%
# ==============================================================================
# BLOQUE 8: VISUALIZACIÓN Y GUARDADO DE GRÁFICOS
# ==============================================================================

import matplotlib.pyplot as plt

try:
    plt.style.use('seaborn-v0_8-darkgrid')
except:
    plt.style.use('seaborn-darkgrid')

# === GRÁFICO 1: Equity Curve ===
fig, ax1 = plt.subplots(figsize=(14, 7))

ax1.plot(motor_dinamico.serie_patrimonio.index, motor_dinamico.serie_patrimonio,
         label='Momentum Crash', color='#ff6b35', lw=2.5, alpha=0.9)

ax1.plot(motor_dinamico.serie_buy_hold.index, motor_dinamico.serie_buy_hold,
         label='Buy & Hold', color='gray', lw=1.5, ls='--', alpha=0.7)

if motor_dinamico.serie_benchmark is not None and not motor_dinamico.serie_benchmark.empty:
    ax2 = ax1.twinx()
    bench_norm = motor_dinamico.serie_benchmark / motor_dinamico.serie_benchmark.iloc[0] * CAPITAL_INICIAL
    ax2.plot(bench_norm.index, bench_norm,
             label=f'{BENCHMARK}', color='#6d4aff', lw=1.8, ls='-.', alpha=0.7)
    ax2.set_ylabel(f'{BENCHMARK} (€)', color='#6d4aff', fontsize=11)
    ax2.tick_params(axis='y', labelcolor='#6d4aff')

ax1.set_xlabel('Fecha', fontsize=12)
ax1.set_ylabel('Patrimonio (€)', color='#ff6b35', fontsize=12)
titulo = f'Momentum Crash — {nombre_evento} ({fecha_ini[:4]}-{fecha_fin[:4]})'
ax1.set_title(titulo, fontsize=14, fontweight='bold')
ax1.legend(loc='upper left', fontsize=11, framealpha=0.9)
ax1.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()

nombre_archivo = f"momentum_{nombre_evento.replace(' ', '_')}_{fecha_ini[:4]}_{fecha_fin[:4]}".lower()
ruta_guardado = Path(gestor_proyecto.DIRS["OUTPUTS_FIGURES"]) / f"{nombre_archivo}.png"
fig.savefig(ruta_guardado, dpi=150, bbox_inches='tight')
print(f"\n✅ Gráfico guardado: {ruta_guardado}")
plt.close()

# === GRÁFICO 2: Drawdown ===
fig2, ax = plt.subplots(figsize=(14, 5))

dd_strategy = (motor_dinamico.serie_patrimonio / motor_dinamico.serie_patrimonio.cummax()) - 1
dd_bench = (motor_dinamico.serie_benchmark / motor_dinamico.serie_benchmark.cummax()) - 1 if motor_dinamico.serie_benchmark is not None else None

ax.fill_between(dd_strategy.index, dd_strategy, 0, color='#ff6b35', alpha=0.35, label='Momentum Crash')
if dd_bench is not None:
    ax.fill_between(dd_bench.index, dd_bench, 0, color='#6d4aff', alpha=0.25, label=f'{BENCHMARK}')

ax.axhline(y=0, color='black', lw=0.5, alpha=0.3)
min_dd_idx = dd_strategy.idxmin()
min_dd_val = dd_strategy.min()
ax.scatter([min_dd_idx], [min_dd_val], s=120, color='red', marker='*', zorder=6,
           label=f'Máximo DD: {min_dd_val:.1%}')

ax.set_title('Evolución del Drawdown — Momentum Crash', fontsize=13, fontweight='bold')
ax.set_ylabel('Drawdown (%)', fontsize=11)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.0%}'))
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11, framealpha=0.9)
fig2.tight_layout()
plt.show()

ruta_dd = Path(gestor_proyecto.DIRS["OUTPUTS_FIGURES"]) / f"{nombre_archivo}_drawdown.png"
fig2.savefig(ruta_dd, dpi=150, bbox_inches='tight')
print(f"✅ Gráfico de drawdown guardado: {ruta_dd}")
plt.close()

print("\n" + "="*70)

# %%
# ==============================================================================
# BLOQUE 9: ANÁLISIS DE CORRELACIÓN
# ==============================================================================

print("\n" + "="*70)
print("🔗 ANÁLISIS DE CORRELACIÓN DEL UNIVERSO MOMENTUM")
print("="*70)

if hasattr(gestor, 'calcular_correlacion'):
    gestor.calcular_correlacion()
    print("\n✅ Análisis de correlación completado.")
else:
    print("⚠️  Método calcular_correlacion no disponible")
    if gestor.precios_ajustados is not None:
        correlaciones = gestor.precios_ajustados.pct_change().corr()
        print(correlaciones.head())

print("\n" + "="*70)
print("📊 ANÁLISIS: DEFENSIVOS vs RIESGOS")
print("="*70)

spy_col = None
tlt_col = None
gld_col = None

for col in gestor.precios_ajustados.columns:
    if 'SPY' in col.upper() or 'SPDR S&P 500' in col:
        spy_col = col
    if 'TLT' in col.upper() or 'Treasury' in col:
        tlt_col = col
    if 'GLD' in col.upper() or 'Gold' in col:
        gld_col = col

if spy_col and tlt_col:
    corr_sp_tlt = gestor.precios_ajustados[spy_col].corr(gestor.precios_ajustados[tlt_col])
    print(f"\n   • Correlación SPY-TLT (total): {corr_sp_tlt:.2f}")
    if corr_sp_tlt < 0:
        print(f"     🟢 Negativa: TLT protege en crisis ✓")

if spy_col and gld_col:
    corr_sp_gold = gestor.precios_ajustados[spy_col].corr(gestor.precios_ajustados[gld_col])
    print(f"   • Correlación SPY-GLD (total): {corr_sp_gold:.2f}")

print("\n" + "="*70)

# %%
# ==============================================================================
# BLOQUE 10: CONCLUSIONES DEL CAPÍTULO 3
# ==============================================================================

print("\n" + "="*70)
print("🎓 CONCLUSIONES DEL CAPÍTULO 3: MOMENTUM CRASH")
print("="*70)
print("""
📌 LECCIONES CLAVE DE ESTE CAPÍTULO
====================================

1. MOMENTUM CRASH SE LOGRÓ IMPLEMENTAR
   • 235 rebalanceos en 19.6 años = 12 operaciones/año
   • CAGR 4.83% vs Contrarian 6.25% (menor, pero aceptable)
   • Max DD -30.6% vs Contrarian -42.3% (MEJOR protección)
   • Downside Capture < 100% = estrategia cumple su propósito

2. LOS FILTROS DE SOBRECOMPRA FUNCIONAN
   • Momentum 12m > 15%: detecta euforia moderada
   • RSI(14) > 70: confirma presión compradora exhausta
   • Juntos filtran burbujas reales de rallies normales

3. LA ROTACIÓN A DEFENSIVOS ES CLAVE
   • Bonos largos (TLT): correlación negativa con riesgos
   • Oro (GLD): hedge contra inflación y pánico
   • Cash (SHV): liquidez para oportunidades futuras
   • Sin defensivos, no hay protección real

4. ESTA ESTRATEGIA NO ES PARA GANAR DINERO
   • CAGR inferior al Buy & Hold (~10%)
   • Su valor está en Downside Capture < 80%
   • Es un SEGURO, no una fuente de alpha primaria
   • Úsala como protección (5-10%), no como inversión

📊 COMPARATIVA FINAL: LOS 3 CAPÍTULOS
======================================

| Aspecto           | Cap. 1 (Core) | Cap. 2 (Contrarian) | Cap. 3 (Momentum) |
|-------------------|---------------|---------------------|-------------------|
| Filosofía         | Buy & Hold    | Comprar miedo       | Vender euforia    |
| Objetivo          | Capturar beta | Alpha defensivo     | Protección        |
| CAGR esperado     | 8-12%         | 6.25%               | 4.83%             |
| Max DD            | -30% a -50%   | -42%                | -31%              |
| Uso en cartera    | 70-80%        | 10-15%              | 5-10%             |

🏆 MODELO CORE-SATELLITE ÓPTIMO
========================================================

CORE (85%):
├── 60%: Indexado global (VTI/VT o 60/40 tradicional)
├── 15%: Contrarian Investing (Cap. 2) — Alpha defensivo
└── 10%: Momentum Crash (Cap. 3) — Seguro anticrisis

SATELLITE (15%):
├── 5%: Factor investing (Value, Quality, Low Vol)
├── 5%: Alternativos (Real Estate, Commodities)
└── 5%: Oportunidades (Crypto, Private Equity)

EXPECTATIVAS DEL MODELO COMBINADO:
• CAGR total: 9-11%
• Max DD: -20% a -30%
• Sharpe ratio: 0.6-0.8
• Downside capture: 40-60%

🎯 RECUERDA
============
El Momentum Crash no es una estrategia para hacer dinero.
Es una estrategia para PERDER MENOS cuando todo el mundo pierde.

En 2008, perder -15% en lugar de -57% te permitió:
• Dormir tranquilo mientras otros quebraban.
• Tener capital para comprar en 2009 a precios de saldo.
• Nunca vender en pánico por necesidad de liquidez.

Eso es alpha verdadero. Eso es lo que construimos aquí.

─────────────────────────────────────────────────────────────────────────
Fin del Capítulo 3: Momentum Crash
Serie completa: 3 capítulos finalizados ✓
─────────────────────────────────────────────────────────────────────────
""")
print("="*70)
print("✅ CAPÍTULO 3 COMPLETADO EXITOSAMENTE")
print("="*70)
```
